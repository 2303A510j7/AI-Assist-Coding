### **2: One-Shot Prompting (Factorial Calculation)**

### #Write a Python function to calculate factorial of a number.Example: Input: 5 → Output: 120

**AI-Generated Code:**

```python
def factorial(n):
    if n < 0:
        return "Invalid input"
    fact = 1
    for i in range(1, n + 1):
        fact *= i
    return fact
```

**Testing:**

input:5->output:120

input:0->output:1

**Observation:**
Better clarity than zero-shot method

### **3:Few-Shot Prompting (Armstrong Number Check)**

 #Write a Python function to check whether a number is an Armstrong number.
Examples: 153 → Armstrong, 370 → Armstrong, 123 → Not Armstrong.

**AI-Generated Code:**

```python
def is_armstrong(num):
    digits = str(num)
    power = len(digits)
    total = sum(int(d)**power for d in digits)
    if total == num:
        return "Armstrong Number"
    else:
        return "Not an Armstrong Number"
```

**Testing:**

input: 153->Armstrong Number

input: 123->Not an Armstrong Number

**Observation:** Few-shot examples improved accuracy and logic of a code

### **4:Context-Managed Prompting (Optimized Number Classification)**

 #Write an optimized Python program to classify a number as Prime, Composite, or Neither with input validation.

**AI-Generated Code**:

```python
def classify_number(n):
    if n <= 1:
        return "Neither Prime nor Composite"
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return "Composite"
    return "Prime"
```

**Testing:**

Input: 2 → Prime

Input: 9 → Composite

Input: 1 → Neither Prime nor Composite

**Observation:**
Optimized and efficient logic.It can solve logical and mathematical solutions

### **5: Zero-Shot Prompting (Perfect Number Check)**

 #Write a Python function to check whether a given number is a perfect number

**AI-Generated Code:**

```python
def is_perfect(num):
    if num <= 0:
        return False
    total = 0
    for i in range(1, num):
        if num % i == 0:
            total += i
    return total == num
```

**Testing:**

Input: 6 → True

Input: 28 → True

Input: 12 → False

**Observation:** Logic correct but inefficient for large numbers

### **6: Few-Shot Prompting (Even or Odd Classification with Validation)**

 #Write a Python program to determine whether a number is even or odd with input validation.
Examples: 8 → Even, 15 → Odd, 0 → Even

**AI-Generated Code:**

```python
def even_or_odd(num):
    if not isinstance(num, int):
        return "Invalid input"
    if num % 2 == 0:
        return "Even"
    else:
        return "Odd"
```

**Testing:**

Input: 8 → Even

Input: 15 → Odd

Input: -4 → Even

Input: 2.5 → Invalid input

**Observation:** Few-shot prompting improved validation and clarity
